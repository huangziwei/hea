### 1.2.1

```python
penguins = data("penguins", package="palmerpenguins")
penguins.glimpse()
```

### 1.2.4

```python
(
    penguins.ggplot(
        x="flipper_length_mm", y="body_mass_g", color="species", shape="species"
    )
    .geom_point()
    .geom_smooth(method="lm")
)
```

```python
(
    penguins.ggplot(
        x="flipper_length_mm",
        y="body_mass_g",
    )
    .geom_point(color="species", shape="species")
    .geom_smooth(method="lm")
    .labs(
        title="Body mass and flipper length",
        subtitle="Dimensions for Adelie, Chinstrap, and Gentoo Penguins",
        x="Flipper length (mm)",
        y="Body mass (g)",
        color="Species",
        shape="Species",
    )
    .scale_color_colorblind()
)
```

### 1.4.1

```python
penguins.ggplot(x="species").geom_bar()
```

```python
from hea.tidy import fct_infreq

penguins.ggplot(x=fct_infreq("species")).geom_bar()
```

### 1.4.2

```python
p1 = penguins.ggplot(x="body_mass_g").geom_histogram(binwidth=20)
p2 = penguins.ggplot(x="body_mass_g").geom_histogram(binwidth=200)
p3 = penguins.ggplot(x="body_mass_g").geom_histogram(binwidth=2000)
p1 + p2 + p3
```

```python
penguins.ggplot(x="body_mass_g").geom_density()
```

### 1.5.1

```python
penguins.ggplot(x="species", y="body_mass_g").geom_boxplot()
```

```python
p1 = penguins.ggplot(x="body_mass_g", color="species").geom_density()
p2 = penguins.ggplot(x="body_mass_g", color="species", fill="species").geom_density(
    alpha=0.5
)
p1 + p2
```

### 1.5.2

```python
p1 = penguins.ggplot(x="island", fill="species").geom_bar()
p2 = (
    penguins.ggplot(x="island", fill="species")
    .geom_bar(position="fill")
    .labs(y="proportion")
)
p1 | p2
```

```python
penguins.ggplot(x="flipper_length_mm", y="body_mass_g").geom_point(
    color="species", shape="species"
).facet_wrap("~island")
```
