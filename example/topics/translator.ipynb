```python
import hea
```

### From hea/Python to R

```python
hea.translate.to_R("""
from hea.models import lm
gala = data('gala', package='faraway')
m0 = lm('Species ~ Area + Elevation + Nearest + Scruz + Adjacent', gala)
m0.summary()
""")
```

```python
hea.translate.to_R("""
from hea.models import lm
gala = data('gala', package='faraway')
m0 = lm('Species ~ Area + Elevation + Nearest + Scruz + Adjacent', gala)
m0.summary()
""", execute=True)
```

### From R to hea/Python

```python
hea.translate.from_R("""
data("gala", package = "faraway")
m0 <- lm("Species ~ Area + Elevation + Nearest + Scruz + Adjacent", gala)
summary(m0)
""")
```

```python
hea.translate.from_R("""
data("gala", package = "faraway")
m0 <- lm("Species ~ Area + Elevation + Nearest + Scruz + Adjacent", gala)
summary(m0)
""", execute=True)
```
